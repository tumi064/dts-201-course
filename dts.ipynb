```python
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

count = len(ages)
print("Number of people:", count)

total = sum(ages)
print("Total:", total)

mean = total / count
print(f"Mean age: {mean:.1f}")

youngest = min(ages)
oldest = max(ages)
print("Youngest:", youngest)
print("Oldest:", oldest)

import statistics
median = statistics.median(ages)
print("Median:", median)
print("Different from mean?", median != mean)
```

```python
import pandas as pd

df = pd.read_excel("50_startups_messy.xlsx")
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())

df = df.drop_duplicates()
df["State"] = df["State"].str.strip().str.title()
df["Marketing Spend"] = pd.to_numeric(df["Marketing Spend"], errors="coerce")

sample_random = df.sample(n=20, random_state=7)
sample_stratified = df.groupby("State", group_keys=False).apply(lambda x: x.sample(frac=0.2, random_state=7))

print(df["R&D Spend"].mean())
print(sample_random["R&D Spend"].mean())
print(sample_stratified["R&D Spend"].mean())

print(df["Marketing Spend"].mean())
print(sample_random["Marketing Spend"].mean())
print(sample_stratified["Marketing Spend"].mean())
```
